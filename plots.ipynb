```python
embedding_dims = [8, 16, 32, 64, 128, 256, 512, 1024]

models = ["GIN", "GINE", "GCNRes"]
layers = [2, 4, 8, 12]

colors = {
    "GIN": "tab:blue",
    "GINE": "tab:orange",
    "GCNRes": "tab:green"
}

linestyles = {
    2: "-",
    4: "--",
    8: "-.",
    12: ":"
}

results = {}

for i, model in enumerate(models):
	dict_aux = {}
	for k, layer in enumerate(layers):
		aux = []
		for j, dim in enumerate(embedding_dims):
			matrix = dict_results[dim]
			# print(matrix)
			# print("layer", layer, k, i)
			aux.append(matrix[k][i])
		# print("---")
		dict_aux[layer] = aux
	results[model] = dict_aux
results
```
